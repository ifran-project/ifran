# Preparation of the environment

## Prerequisites
* You have a Slices account 
* You have a basic understanding of the slices CLI


## Step 1 - Register to SLICES

The Post-5G blueprint is carried out within the SLICES framework, where resources
are structured into *projects* and *experiments*. Each experiment is associated
with a project. Make sure you have a valid SLICES-RI account and that you are
registered to a project. Visit the [SLICES Poral](https://portal.slices-ri.eu/)
for more details.


## Step 2 - Install and configure the SLICES-CLI
Install the [Command Line Interface (SLICES-CLI)](https://doc.slices-ri.eu/SupportingServices/slicescli.html),
log-in with your SLICES account and set the project to use (e.g., `post5g-beta`).



```python
! pip install --upgrade slices-cli --extra-index-url=https://doc.slices-ri.eu/pypi/
! slices auth login
```

Select your project with the following command (adapt the project name to yours).

```python
! slices project use post5g-beta
```


If it is not done already, [register your public SSH key](https://doc.slices-ri.eu/SupportingServices/slicescli.html#registering-your-public-ssh-key)
to SLICES. These keys are automatically installed on the resources used in
with the Post-5G blueprint.

This initialization process is required only when you install the SLICES-CLI.

You are now ready to run your experiment!
